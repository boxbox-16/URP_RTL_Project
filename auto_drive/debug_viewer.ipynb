```python
import zmq
import numpy as np
import ipywidgets as widgets
from IPython.display import display

PORT = 5556   # default.yaml debug_stream.port 와 맞춰야 함

w = widgets.Image(format='jpeg', width=640, height=240)
display(w)

ctx = zmq.Context()
sub = ctx.socket(zmq.SUB)
sub.connect(f"tcp://localhost:{PORT}")
sub.setsockopt(zmq.SUBSCRIBE, b"")
sub.setsockopt(zmq.RCVTIMEO, 2000)

print(f"[Viewer] 연결 중 localhost:{PORT} — 메인 노트북 파이프라인을 먼저 실행하세요")
print("Interrupt Kernel (■) 버튼으로 종료")

frame_count = 0
try:
    while True:
        try:
            w.value = sub.recv()
            frame_count += 1
        except zmq.Again:
            print(f"대기 중... (수신 {frame_count}프레임)", end='\r')
except KeyboardInterrupt:
    pass
finally:
    sub.close()
    ctx.term()
    print(f"\n[Viewer] 종료. 총 수신: {frame_count}프레임")
```
